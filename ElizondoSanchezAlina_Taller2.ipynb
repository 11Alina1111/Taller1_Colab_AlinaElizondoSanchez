```python
# Linealidad
#Operador= L(x) = 2x
#Vectores
u = [1., 2.]
v= [3., -1.]

#Escalares
alpha= 2
beta=-1

# Calacular lado derecho
izquierda = [
    2 * (alpha*u[0] + beta*v[0]),
    2 * (alpha*u[1] + beta*v[1])]

# Calcular lado derecho:
derecha = [alpha*(2*u[0]) + beta*(2*v[0]),
           alpha*(2*u[1]) + beta*(2*v[1])]

print("Izquierda:", izquierda)
print("Derecha:", derecha)
print("Son lineales?", izquierda == derecha)
```
