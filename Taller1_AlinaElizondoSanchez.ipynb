### B.1.Estadística descriptiva

```python
# Estadistica descriptiva
print ('Descripción:\n', df.describe())

#Verificar columnas numéricas
numericas =df.select_dtypes(include=["number"]).columns
print("Columnas numéricas:", list(numericas))
print(df.dtypes)

print('\n')
#Age
print('EDAD')
print('Edad promedio', (df['age'].mean()))
print ('Mediana de edad:',(df['age'].median()))
print ('Desviación estándar de edad', ( df['age'].std()))
print ('\n')

#Presión
print ('PRESIÓN')
print('Presión promedio', df['resting_blood_pressure'].mean())
print('Mediana de presión', df['resting_blood_pressure'].median())
print('Desviación estándar', df['resting_blood_pressure'].std())
print('\n')

#Colesterol
print ('COLESTEROL')
print ('Promedio', df['cholestoral'].mean())
print ('Mediana', df['cholestoral'].median())
print ('Desviación estándar', df['cholestoral'].std())
print('\n')

print('THALASSEMIA')
#df['thalassemia']= df['thalassemia'].astype('category')
#df['thalassemia'].mean().astype()
#df['thalassemia'].median()
#df['thalassemia'].std()
print('\n')

#oldpeak
print ('OLDPEAK')
print ('Promedio:', df['oldpeak'].mean())
print('Mediana:', df['oldpeak'].median())
print('Desviación estándar', df['oldpeak'].std())
print('\n')

# Variables categoricas
# variables_categoricas = df.groupby['thalassemia']('target')[variables].agg(lambda x: x.mode()[0])
# print(variables_categoricas)

# Convertir talasia a numerico
df['thalassemia'] = pd.to_numeric(df['thalassemia'], errors='coerce')

# Variables numéricas
variables = ['age', 'resting_blood_pressure', 'cholestoral', 'thalassemia', 'oldpeak']


#Estadística descriptiva
print('Estadística descriptiva mediante Target\n')
agrupado = df.groupby('target')[variables].agg(['mean', 'median', 'std'])
print(agrupado)

##Interpretación
#### Primero se procedió con un describe() que permite dar un resumen del promedio
#### la mediana y la desviación estándar de cada variable. Esto muestra de
#### talasemia posee NaN, esto porque es cualitativa. Para verificarlo se procede
#### a visualizar sólo las columnas numéricas, en la que claramente no aparece
#### talasmia. Así que mediante, mean(), median() y std() se procede a calcular
#### el promedio, la mediana y la desviación estándar de cada variable numérica.
#### Pero, para que Thalassemia no quede fuera de la observación se convierte en
#### numérico. Ahora teniendo todas la variables convertidas a numéricas,
#### se agrupan por groupby(), para obtener el target con el promedio, medaina y
#### desviación estándar de cada variable.
#### No obstante, se debe recordar que lo ideal es mantener las variables de su
#### forma original. Para generar datos más prácticos.
#### En este caso, las variables cuantitativas, se utiliza medidas de
#### tendencia central, algo que en variables cualitativas no se utiliza.
```

## Parte C. Identificar Outliers y Matriz de Correlación de Pearson.
Identifique outliers rápidos en las variables target seleccionadas en el apartado anterior (parte B).
Cree un gráfico de matriz de correlación, seleccionando los campos o variables del punto anterior.

### C.1.Identificar outliners

```python
# Identificar outlinier

#1_Cuartiles
Q1 = df[variables].quantile(0.25)
Q3 = df[variables].quantile(0.75)

#2_Rango intercuartilico
IQR = Q3 - Q1

#3_ Contar outliners
outliers = ((df[variables] < (Q1- 1.5 * IQR)) | (df[variables] > (Q3 + 1.5 * IQR))).sum()
print('Cantidad de outliniers por variables:\n', outliers)
print('\n')

#Otra forma es calcular outliners con límite predeterminado
print('Outliers con límite:')

#1_Cuartiles
q1 = df[variables].quantile(0.25)
q3 = df[variables].quantile(0.75)
#2_Calculo de rango intercuartilico
iqr = q3 - q1
#3_limite para detectar outliners
lower_cut = q1 - 1.5*iqr
#4_ contar outliners
outliers = df[df[variables] < lower_cut]
print("Outliers en variables:")
print(outliers[variables])


###Interpretación
### De la primera forma se encuentran outliners en:
### la presión arterial en reposo con 30,
### el colesterol con 16 y la depresión posee 7.
### Los outiliners son fundamentales, para poder verificar si son datos erroneos
### o si en este caso se trata de casos donde el paciente posee valores atípicos,
### por ejemplo, en su nivel de colesterol y debe ser tratado para regular esos
### valores.
#### Lugo, los outliners con valores límites permiten ser más objetivos, en cada
#### variable, por ejemplo, establecer un nivel de glucosa que no sea menor
####vo superior, según datos médicos. No obstante, para este caso, se usa el 1.5
#### como valor estándar. En este caso, todos los valores de NaN, lo cual
#### indicaría que no se encontraron valores atípicos.
```

### C.2.Matriz de correlación

```python
# Calcular correlación
correlacion = df[variables].corr()
print('Correlación de variables\n')
print(correlacion)
print('\n')

# Matriz de correlación
print('Gráfico 1. Matriz de correlación')
plt.figure(figsize=(12, 6))
sns.heatmap(correlacion, annot=True, cmap='crest', vmin=-1, vmax=1)
plt.title('Matriz de correlación')
plt.show()

###Interpretación
##Correlación: Cuando el valor es positiva, la correlación es fuerte, cuando es
##negativa la correlación es débil y si es igual a cero, no tiene correlación.
##Por ejemplo, la relación de edad con la presión arterial es de 0.27, lo cual
## indica correlación fuerte. Entre más edad mayor presión arterial en reposo.
## El colesterol con la depresión es una correlación débil de 0,06.
## Mientras, que thalassemia no tiene relación( NaN), esto porque es un tipo
## de dato cualitativo.

## Matriz de correlación: En esta matriz,thalassemia no posee color, porque no
## tiene relación al ser un dato cualitativo.
#Colesterol & presión arterial de reposo: Posee una correlación débil de 0.13.
##Edad & Depresión: Correlación débil de 0,21.
##Edad & Presión arterial en reposo: Presenta la correlación más fuerte con
## un 0.27
#Depresión & colesterol: Presenta la correlación más débil con un 0.06,
### casi nula, porque está muy cerca del cero.
```
